```python
import numpy as np
import pandas as pd
import matplotlib
from matplotlib import pyplot as plt

COLORS = matplotlib.rcParams['axes.prop_cycle'].by_key()['color']
```

```python
data = pd.read_csv('data.csv').to_numpy()
t = np.take(data, [2], axis=1)
y = np.take(data, [3,4], axis=1)
y_pred = pd.read_csv('cpp/pred.csv', header=None).to_numpy()
```

```python
label = ['x [m]', 'v [m]']
plt.plot([], [], c='k', label='True')
plt.plot([], [], '^--', c='k', label='C++ Pred')
for i in range(2):
    plt.plot(t, y[:,i], c=COLORS[i], lw=1.)
    plt.plot(t, y_pred[:,i], c=COLORS[i], ls='--', marker='^', markevery=5, lw=1.)
    plt.plot([], [], c=COLORS[i], label=label[i])
plt.xlabel('t [s]')
plt.legend()
# plt.savefig('./inference.pdf')
```
