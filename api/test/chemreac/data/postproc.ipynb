```python
import numpy as np
import pandas as pd
import matplotlib
from matplotlib import pyplot as plt

COLORS = matplotlib.rcParams['axes.prop_cycle'].by_key()['color']
```

```python
def plot_true_pred(x, y, y_pred, label, figname):
    plt.figure()
    plt.plot([], [], c='k', label='True')
    plt.plot([], [], '^--', c='k', label='Pred')
    for i in range(2):
        plt.plot(t, y[:,i], c=COLORS[i], lw=1.)
        plt.plot(t, y_pred[:,i], c=COLORS[i], ls='--', marker='^', markevery=100, lw=1.)
        plt.plot([], [], c=COLORS[i], label=label[i])
    plt.xscale('log')
    plt.xlabel('t [s]')
    plt.legend()
    plt.savefig(figname)
    plt.close()
```

```python
data = pd.read_csv('true.csv').to_numpy()
t = np.take(data, [4], axis=1)
y = np.take(data, [5,6,7,8], axis=1)
y_pred = np.loadtxt('pred.csv')
```

```python
for k in ('mass','temp'):
    idx = [0,1] if k == 'mass' else [2,3]
    if k == 'mass':
        idx = [0,1]
        label = ['$Y_{\text{O}_2}$', '$Y_{\text{O}}$']
    else:
        idx = [2,3]
        label = [r'$T_h$', r'$T_v$']    
    plot_true_pred(t, np.take(y, idx, 1), np.take(y_pred, idx, 1), label, f'./{k}.pdf')
```
