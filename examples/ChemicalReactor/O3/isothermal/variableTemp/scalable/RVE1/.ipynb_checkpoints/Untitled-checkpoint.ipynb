```python

graph = [0,0,0,0]

graph[0] = [0]

graph[1] = {
    0: [0,1,2]
}

graph[2] = {
    0: {
        0: [0,1,2],
        1: [3,4,5],
        2: [6,7,8]
    }
}

graph[3] = {
    1: {
        1: {
            1: [1,2,3],
            2: [4,5,6],
            3: [7,8,9]
        },
        2: {
            4: [10,11,12],
            5: [13,14,15],
            6: [16,17,18]
        },
        3: {
            7: [19,20,21],
            8: [22,23,24],
            9: [25,26,27]
        }
    }
}


for i in graph:
    print(depth(i)+1)

def con(g, level=0, conn={}):
    d = {}
    conn[level] = []
    for i in g.values():
        if hasattr(i, 'keys'):
            l = list(i.keys())
            d = {**d, **i}
        else:
            l = i
        conn[level].append(l)
    return con(d, level=level+1, conn=conn) if d != {} else conn
    
ll_conn = con(graph[2])
```

```python
for i in graph:
    ts = time_scales(i)
    print(ts)
```
