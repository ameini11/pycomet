```python
import h5py
import tensorflow as tf
```

```python
case_dir = '../../ControlONet/trained/delta/'
```

```python
def rename(f, i, j, macro):
    # Old group
    old_gr = macro + j
    # Renaming
    try:
        new_gr_ = f'/output_fn_NEQ3_opnn_{i}'
        old_gr_ = f'/output_fn_NEQ3_opnn_{j}'
        f[old_gr + new_gr_] = f[old_gr + old_gr_]
        del f[old_gr + old_gr_]
    except:
        pass
    new_gr = macro + str(i)
    f[new_gr] = f[old_gr]
    del f[old_gr]
```

```python
group = '/NEQ3/opnn_'
j = '1'
for i in (2, 3):
    p = case_dir + f'./saving/delta3{i}.h5'
    f = h5py.File(p,'r+')
    
    # OpNN ------------------------------------------------
    macro = 'NEQ3' + group
    rename(f, i, j, macro)
    
    # Output transformation -------------------------------
    try:
        macro = 'output_fn_NEQ3_opnn_' + j + group
        rename(f, i, j, macro)
        old_gr = f'/output_fn_NEQ3_opnn_{j}'
        new_gr = f'/output_fn_NEQ3_opnn_{i}'
        f[new_gr] = f[old_gr]
        del f[old_gr]
    except:
        macro = f'output_fn_NEQ3_opnn_{i}' + group
        rename(f, i, j, macro)
    
    f.close()
```
