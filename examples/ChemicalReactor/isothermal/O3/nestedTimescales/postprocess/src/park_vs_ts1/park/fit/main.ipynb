## QSS Dissociation Rate
The QSS rates are obtained as a weighted sum of the state specific dissociation rates, with the weights given by the QSS population distribution:
\begin{equation}
K_{QSS}=\sum_ik_i^D(T)\frac{n_i}{n}
\end{equation}

## Characteristic internal energy relaxation time: OPTION 1
In order to calculate the characteristic internal energy relaxation time time, the initial portion of the vibrational temperature history is fit using the Landau-Teller expression,
\begin{equation}
    \epsilon_{int}\left(t\right) = \epsilon^*_{int} − \Delta\epsilon_{int}\left(exp\left(−\frac{t}{\tau}\right)\right)
\end{equation}
Here
- $\epsilon^*_{int}$ is the final steady-state average vibrational energy,
- $\Delta\epsilon_{int}$ is the difference between the initial and final average internal energies,
- $τ$ is the characteristic vibrational relaxation time.
