Import libraries

```python
import os
```

Define input temperature

```python
T = 10000
```

Define paths and file names

```python
file_names = ['Diss_Corrected', 'Exch_Type1', 'Inel']
out_file_name = f'O3_fullyStS_T{T}K'
cwd = os.getcwd()
```

Merge files

```python
with open(f'{cwd}/{out_file_name}', 'w') as f:
    f.write('Units=cm^3/s')
    for fname in file_names:
        with open(f'{cwd}/T{T}K/{fname}.dat') as infile:
            for line in infile:
                f.write(line)
```
