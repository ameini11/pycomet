Recursive inference solver

```python
def recursive_inference(
    ic,
    dt0=1.e-7,
    t_end=1.e-3,
    dt_str=1.4,
    dt_max=1.e-4
):
    # Initialize variables
    T = np.array([[0.]])
    Y = tf.concat(
        [1. - np.sum(ic[0:1,2:], keepdims=True), ic[0:1,2:]], axis=1
    )
    # Initialize loop
    t = T
    dt = dt0
    while t <= t_end:
        # Update inputs
        ic_ = np.expand_dims(tf.concat([ic[0,:2], Y[-1,1:]], axis=0), axis=0)
        t_ = np.array([[dt]])
        # Predictions
        y = output_ops(model.net([ic_,t_]), timescale=3)
        # Update time step
        t = t + dt
        dt = min(dt*dt_str, dt_max)
        # Collect data
        T = np.vstack((T, t))
        Y = np.vstack((Y, y))
    return T, Y
```

```python
def plot_recursive_inference(
    path,
    x,
    y,
    x_pred,
    y_pred,
    var_names=None,
    fig_name=None,
    labels=[r'$t$ [s]', r'$Y$'],
    scales=['log', 'linear'],
    xlim=None,
    ylim=None
):
    fig = plt.figure()

    # Set up figure
    x_label, y_label = labels
    x_scale, y_scale = scales

    # X axis
    if x_label is not None:
        plt.xlabel(x_label)
    if x_scale is not None:
        plt.xscale(x_scale)
    if xlim is not None:
        plt.xlim(xlim)
    else:
        plt.xlim([min(x_pred), max(x_pred)])

    # Y axis
    if y_label is not None:
        plt.ylabel(y_label)
    if y_scale == 'log':
        plt.yscale(y_scale)
        if ylim is None:
            plt.ylim([np.amin(y)*5.e-1, np.amax(y)*5.e+0])
    else:
        delta = np.amax(y)*0.1
        if ylim is None:
            plt.ylim([np.amin(y)-delta, np.amax(y)+delta])
    if ylim is not None:
        plt.ylim(ylim)

    if len(y.shape) < 2:
        y = np.expand_dims(y, 1)

    # Plot figure
    # Define styles
    true_style = dict(
        lw=1.
    )
    pred_style = dict(
        marker='x',
        fillstyle='none',
        markersize=5,
        ls='',
        lw=1.
    )

    # Define parameters
    colors = []
    n = y.shape[1]

    # Loop over features
    for d in range(n):

        # >> Plot true values
        plt.plot(
            x,
            y[:,d],
            **true_style
        )
        colors.append(plt.gca().lines[-1].get_color())

        # >> Plot predicted values
        plt.plot(
            x_pred,
            y_pred[:,d],
            **pred_style,
            label=var_names[d],
            c=colors[-1]
        )

    # Plot color legends
    if n > 12:
        fs = 'x-small'
    elif n > 6:
        fs = 'small'
    else:
        fs = 'medium'
    plt.legend(
        [plt.plot([], [], c=colors[i])[0] for i in range(len(var_names))],
        var_names,
        fontsize=fs,
        ncol=int(np.ceil(n/3))
    )

    # Save figure
    fig_path = path + '/' + fig_name + '.pdf'
    fig.savefig(fig_path)
    plt.close()
```

## Nonequilibrium distance
A linear form for the distribution function, which results in a thermalized local Boltzmann distribution within individual bins, has been employed.

This function is defined as follows
\begin{equation*}
    \mathcal{F}_i\left(\epsilon_s\right):\qquad\log\left(\frac{g_s}{n_s}\right)=\alpha_i+\beta_i\epsilon_s
    \quad\forall s\in\mathcal{I}_i
\end{equation*}
where
\begin{align}
    \alpha_i&=\log\left(\frac{Q_i}{n_i}\right)\label{eq:alpha}\\
    \beta_i&=\frac{1}{k_BT_i}\nonumber\\
    n_i&=Y_i\frac{\rho}{m}\nonumber
\end{align}

Setting up

```python
path_ = path + '/noneq/'
if not os.path.exists(path_):
    os.makedirs(path_)

# Associate Timescale to grouping
ts_to_grouping = {
    'TS1': 'RVE1',
    'TS2': 'RVE3',
    'TS3': 'RVE9',
    'TS4': 'RVE27'
}
```

Species thermodynamics

```python
def get_species_thermos(
    ts_to_grouping,
    species
):
    ts_to_species_thermos = {}
    for timescale, grouping in ts_to_grouping.items():
        # Update O2 with the proper grouping strategy
        species['O2']['rovib']['grouping'] = grouping
        # Initialize species thermodynamics
        spTh = {
            name: specieMdl.specieThermos.SpecieThermos(
                name,
                info,
                thermo='CoupledEnergyModes',
                EOS='PerfectGas',
            ) for name, info in species.items()
        }
        # Add species thermodynamics to the corresponding timescale
        ts_to_species_thermos[timescale] = spTh
    return ts_to_species_thermos

def update_partition_functions(
    ts_to_species_thermos,
    T
):
    Q = []
    for ts, species_thermos in ts_to_species_thermos.items():
        # Update species thermodynamics
        for spTh in species_thermos.values():
            spTh.intPF.update(T)
        # Get internal partition function
        Q.append(species_thermos['O2'].intPF.Q)
    return Q
```

```python
# Get species thermodynamics
ts_to_species_thermos = get_species_thermos(
    ts_to_grouping,
    inputs.general.species
)
```

### Nonequilibrium distance definition: OPTION 1
Consider the zero order truncation of the Maximum Entropy Principle reconstruction
\begin{align*}
    f_i=\alpha_i&=\log\left(\frac{Q_i}{n_i}\right)\\
    &=\log\left(\frac{Q_i}{Y_i}\frac{m}{\rho}\right)
\end{align*}

```python
def eval_alpha(rho, Y, Q):
    return tf.nest.map_structure(
        lambda q, y: np.log((q/y)*(2*const.mass['O']/rho)), Q, Y
    )

def get_alpha(data):
    # Extract data
    (ic, t), Y = data

    # Get temperature and density from initial condition
    T = np.take(ic, [0], axis=1)
    rho = np.take(ic, [1], axis=1)

    # Update partition functions with the temperature of the physical scenario
    Q = update_partition_functions(ts_to_species_thermos, T)

    # Extract only O2 mass fraction from Timescale 1
    Y[0] = np.take(Y[0], [0], axis=1)
    
    return t, eval_alpha(rho, Y, Q)
```

Nonequilibrium distance calculated as norm $\left(p=1\right)$ of the differential offset
* TS$^{(3)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_i^{(3-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_i^{(3)}}
    \left\lVert f_i^{(3)}-f_j^{(4)}\right\rVert_p
    \quad\forall\mathcal{I}_i^{(3)}\in\text{RVE}_9
\end{equation*}
* TS$^{(2)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_i^{(2-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_i^{(2)}}
    \left\lVert f_i^{(2)}-f_j^{(4)}\right\rVert_p
    \quad\forall\mathcal{I}_i^{(2)}\in\text{RVE}_3
\end{equation*}
* TS$^{(1)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_1^{(1-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_1^{(1)}}
    \left\lVert f_1^{(1)}-f_j^{(4)}\right\rVert_p
    \quad\mathcal{I}_1^{(1)}\in\text{RVE}_1
\end{equation*}

```python
def eval_dalpha_norm(alpha, norm_order=2):
    dalpha_norm = []
    for ts, alpha_ts in enumerate(alpha):
        dalpha_norm_ts = []
        if ts == len(alpha)-1:
            continue
        indeces = model.net.ts_fast_conn[ts]
        for i, idx in enumerate(indeces):
            alpha_fast = np.take(alpha[-1], idx, axis=1)
            alpha_slow = np.take(alpha_ts, [i], axis=1)
            norm = tf.norm(
                alpha_slow - alpha_fast, ord=norm_order, axis=1, keepdims=True
            )
            dalpha_norm_ts.append(norm)
        dalpha_norm.append(tf.concat(dalpha_norm_ts, axis=1))
    return dalpha_norm
```

### Nonequilibrium distance definition: OPTION 2
Consider the first order truncation of the Maximum Entropy Principle reconstruction
\begin{align*}
    f_i(\epsilon_s)=\alpha_i+\beta_i\epsilon_s&=\log\left(\frac{Q_i}{n_i}\right)+\frac{\epsilon_s}{k_BT_i}\\
    &=\log\left(\frac{Q_i}{Y_i}\frac{m}{\rho}\right)+\frac{\epsilon_s}{k_BT_i}
\end{align*}

```python
def eval_beta_eps(T):
    E = ts_to_species_thermos['TS1']['O2'].specie.rv_lev['E']
    T_s = (E / const.EV_to_J) / const.UKB
    beta_eps_ = tf.linalg.matmul(1./T, T_s, transpose_b=True)
    # Loop over timescales
    beta_eps = []
    for ts, species_thermos in ts_to_species_thermos.items():
        beta_eps_ts = []
        sp = species_thermos['O2'].specie
        for bin_i in range(sp.n_bins):
            mask = sp.lev_to_bin == bin_i
            beta_eps_ts.append(tf.boolean_mask(beta_eps_, axis=1))
        beta_eps.append(beta_eps_ts)
    return beta_eps

def mep_reconstruction(data, order=1):
    # Extract data
    (ic, t), Y = data

    # Get temperature and density from initial condition
    T = np.take(ic, [0], axis=1)
    rho = np.take(ic, [1], axis=1)

    # Update partition functions with the temperature of the physical scenario
    Q = update_partition_functions(ts_to_species_thermos, T)

    # Extract only O2 mass fraction from Timescale 1
    Y[0] = np.take(Y[0], [0], axis=1)

    # Eval MEP reconstruction
    alpha = eval_alpha(rho, Y, Q)
    if order == 0:
        f = alpha
    elif order == 1:
        f = []
        beta_eps = eval_beta_eps(T)
        for i in range(len(alpha)):
            alpha_i = tf.split(alpha[i], alpha[i].shape[1], axis=1)
            f_i = tf.nest.map_structure(
                lambda a, be: a+be, alpha_i, beta_eps[i]
            )
            f.append(tf.concat(f_i, axis=1))
    return t, f
```

Nonequilibrium distance calculated as norm $\left(p=1\right)$ of the differential offset
* TS$^{(3)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_i^{(3-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_i^{(3)}}
    \left\lVert f_i(\epsilon_s)^{(3)}-f_j(\epsilon_s)^{(4)}\right\rVert_p
    \quad\wedge s\in\mathcal{I}_j^{(4)}
    \qquad\forall\mathcal{I}_i^{(3)}\in\text{RVE}_9
\end{equation*}
* TS$^{(2)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_i^{(2-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_i^{(2)}}
    \left\lVert f_i(\epsilon_s)^{(2)}-f_j(\epsilon_s)^{(4)}\right\rVert_p
    \quad\wedge s\in\mathcal{I}_j^{(4)}
    \qquad\forall\mathcal{I}_i^{(2)}\in\text{RVE}_3
\end{equation*}
* TS$^{(1)}$-to-TS$^{(4)}$
\begin{equation*}
    \epsilon_1^{(1-4)} = \sum_{j,\mathcal{I}_j^{(4)}\in\mathcal{I}_1^{(1)}}
    \left\lVert f_1(\epsilon_s)^{(1)}-f_j(\epsilon_s)^{(4)}\right\rVert_p
    \quad\wedge s\in\mathcal{I}_j^{(4)}
    \qquad\mathcal{I}_1^{(1)}\in\text{RVE}_1
\end{equation*}

```python
def eval_df_norm(f, norm_order=1):
    df_norm = []
    for ts, f_ts in enumerate(f):
        df_norm_ts = []
        if ts == len(f)-1:
            continue
        indeces = model.net.ts_fast_conn[ts]
        for i, idx in enumerate(indeces):
            f_fast = np.take(f[-1], idx, axis=1)
            f_slow = np.take(f_ts, [i], axis=1)
            norm = tf.norm(
                f_fast - f_slow, ord=norm_order, axis=1, keepdims=True
            )
            df_norm_ts.append(norm)
        df_norm.append(tf.concat(dalpha_norm_ts, axis=1))
    return df_norm
```

Compute the levels distribution dynamics for each timescale

```python
def get_ts_lev_dist(data):
    # Extract data
    (ic, t), Y = data

    # Get temperature and density from initial condition
    T = np.take(ic, [0], axis=1)
    rho = np.take(ic, [1], axis=1)

    # Extract O and O2 mass fraction from Timescale 1
    Y_O = np.take(Y[0], [1], axis=1)
    Y[0] = np.take(Y[0], [0], axis=1)

    # Loop over the various timescale mixtures
    ts_lev_dist = []
    mix = dict(O=Y_O)
    for i, (ts, mix_ts) in enumerate(ts_to_mixture.items()):
        mix['O2'] = Y[i]
        # Update the mixture with the new mass fractions
        mix_ts.update(mix)
        mix_ts.n_(rho=rho)
        # Update the partition functions
        for spTh in mix_ts.spTh.values():
            spTh.intPF.update(T)
        # Get n, g, E levels for each bin
        ts_lev_dist.append(
            mix_ts.spTh['O2'].intPF.levels_distribution()
        )
    return t, ts_lev_dist
```

Coarse-grained levels distribution plotter

```python
def plot_levels_dist(
    lev_dist=None,
    var_names=lev_dist_names,
    time_instant=None,
    title=None,
    labels=[r'$\epsilon_i$ [eV]', r'$n_i\//\/g_i$ [m$^{-3}$]'],
    scales=['linear', 'log'],
    xlim=None,
    ylim=None,
    fig_ax=None,
    save=True,
    path=None
):
    if fig_ax is None:
        fig, ax = plt.subplots()
    else:
        fig, ax = fig_ax

    # Set up figure -----------------------------------------------------------
    if title is not None:
        ax.title(title)
    x_label, y_label = labels
    x_scale, y_scale = scales
    # X axis
    if x_label is not None:
        ax.xlabel(x_label)
    if x_scale is not None:
        ax.xscale(x_scale)
    if xlim is not None:
        ax.xlim(xlim)
    # Y axis
    if y_label is not None:
        ax.ylabel(y_label)
    if x_scale is not None:
        ax.yscale(y_scale)
    if ylim is not None:
        ax.ylim(ylim)

    # Plot figure -------------------------------------------------------------
    style = dict(
        lw=1,
        marker='o',
        fillstyle='full',
        markersize=0.3,
        linestyle='none'
    )
    # Loop over Timescales
    for i, (n, g, E) in enumerate(lev_dist):
        # Loop over bins
        for b in range(len(n)):
            n_b = tf.reshape(n[b][idx,:], (-1,1))
            ax.plot(E[b], n_b/g[b], c=colors[i], **style)
    # Time instant
    if time_instant is not None:
        ax.text(
            0.75, 0.92,
            r'$t$ = %.1e s' % time_instant,
            transform=ax.transAxes,
            fontsize=14
        )
    # Add legend
    if var_names is not None:
        ax.legend(
            [ax.plot([], [], c=colors[i])[0] for i in range(len(var_names))],
            var_names
        )

    # Save figure -------------------------------------------------------------
    if save:
        fig.savefig(path)
        
    if fig_ax is not None:
        return fig, ax
```
