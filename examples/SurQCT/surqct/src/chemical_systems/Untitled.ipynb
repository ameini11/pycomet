```python
import numpy as np
from scipy.interpolate import UnivariateSpline
import matplotlib.pyplot as plt
x = np.arange(100)/100
y = np.sin(x)
cs = UnivariateSpline(x, y, k=3, s=0)
```

```python
r = cs.roots()
```

```python
r.squeeze()
```

```python
plt.plot(x, y)
plt.scatter(r, [0.]*len(r))
```
