```python
# Paths
WORKSPACE_PATH   = '/home/zanardi/WORKSPACE/'
PathToRunFldr    = WORKSPACE_PATH + '/ML/PyCOMET/pycomet/examples/SurQCT/O3/inel/'
PathToOutputFldr = WORKSPACE_PATH + '/AirDatabase/Run0D/database/levels/samples/Uniform/'
```

```python
# Modules
sys.path.insert(0, PathToRunFldr + '/data_gen/')
import InputData
```

```python
sys.path.insert(0, InputData.SurQCTFld)
from surqct.src.Utils import Paths, Reading
```

```python
# Temperature vector
TTranVec = InputData.TTranVecTrain
NTTran   = len(TTranVec)

# Reading Levels Info of Initial and Final Molecules
DiatData = {}
NLevels  = {}
for iMol, (Mol, Sys) in enumerate(InputData.Molecules):
    if Mol not in DiatData:
        DiatData[Mol] = Reading.read_diatdata(Paths.DiatFile[Mol][Sys], Mol, TTranVec)
        NLevels[Mol] = DiatData[Mol].shape[0]
```

```python
NSamples = 5000

# Sampling levels uniformly
for iT in range(NTTran):
    for iMol, (Mol, Sys) in enumerate(InputData.Molecules):
        iDxVec = np.linspace(1, NLevels[Mol], num=NSamples, dtype=int)
        iDxVec = pd.DataFrame(iDxVec, columns=['iLevel'])
        iDxVec.to_csv(PathToOutputFldr+Mol+'_'+str(NSamples)+'_'+str(int(TTranVec[iT]))+'K.csv', index=False)
```
