Plotting

```python
import matplotlib
from matplotlib import pyplot as plt
COLORS = matplotlib.rcParams['axes.prop_cycle'].by_key()['color']

path_to_figs = "./figs/"
os.makedirs(path_to_figs, exist_ok=True)

def plot_var(
    x=None,
    y=None,
    xref=None,
    yref=None,
    labels=[r'$t$ [s]', r'$X$'],
    scales=['log', 'log'],
    names=None,
    figname=None
):
    # Initialize figures
    fig = plt.figure()
    ax = fig.add_subplot()
    # x axis
    ax.set_xlabel(labels[0])
    ax.set_xscale(scales[0])
    # y axis
    ax.set_ylabel(labels[1])
    ax.set_yscale(scales[1])
    # Plotting
    n, d = y.shape if (x is not None) else yref.shape
    true_style = dict(
        lw=1
    )
    pred_style = dict(
        marker='^',
        fillstyle='none',
        markersize=5,
        markevery=n//20,
        ls='--',
        lw=1
    )
    for i in range(d):
        if (x is not None):
            ax.plot(x, y[:,i], c=COLORS[i], **pred_style)
        if (xref is not None):
            ax.plot(xref, yref[:,i], c=COLORS[i], **true_style)
        ax.plot([], [], c=COLORS[i], label=names[i])
    # plt.ticklabel_format(axis='y', style='sci', scilimits=(0,0))
    plt.legend()
    if (figname is not None):
        fig.savefig(figname)
    plt.show()
    plt.close()
```

Construct rate matrices

```python
ke = 0.0
for reaction in kin.reactions:
    if (reaction.param['process'] in (5,6)):
        ke = ke + reaction.fwd.k.numpy().squeeze()[0]
        ke = ke + reaction.bwd.k.numpy().squeeze()[0].T
    elif (reaction.param['process'] == 2):
        kd = reaction.fwd.k.numpy().squeeze()[0]
        kr = reaction.bwd.k.numpy().squeeze()[0]
ke = (ke.T - np.diag(np.sum(ke, axis=-1)))
```

Define integrals
> Full system

```python
def ful_fun(t, c):
    f = np.zeros_like(c)
    # Extract number densities
    n = c * const.UNA
    na, nm = n[0:1], n[1:]
    # Construct source terms
    we = np.matmul(ke,nm*na)
    wd = kd*nm*na - kr*na**3
    # Compose source terms
    f[0] = 2*np.sum(wd)
    f[1:] = we - wd
    return f / const.UNA

def ful_int(t, n0):
    res = sp.integrate.solve_ivp(
        fun=ful_fun,
        t_span=[t[0],t[-1]],
        t_eval=t.squeeze(),
        y0=n0/const.UNA,
        method='LSODA',
        atol=1e-9,
        rtol=1e-7
    )
    return res.y.T * const.UNA
```

```python
n_full = ful_int(t, n[0])
```

```python
plot_var(
    x=t.squeeze()[1:],
    y=n_full[1:],
    xref=t.squeeze()[1:],
    yref=n[1:],
    labels=[r'$t$ [s]', r'$n$ [m$^{-3}$]'],
    scales=['log', 'log'],
    names=['O'] + [r'O$_2^{(%d)}$' % (i+1) for i in range(3)],
    figname=path_to_figs + "/solution"
)
```

> Linear system

```python
def exc_int(t, na, nm):
    ek = sp.linalg.expm(t*ke*na)
    nm = ek @ nm
    return na, nm
```

> Forced linear system

```python
def exc_dis_int(t, na, nm):
    # Construct source terms
    wr = kr*na**3
    # Construct solution operator
    ek = sp.linalg.expm(t*ke*na)
    ki = np.linalg.pinv(ke*na)
    i = np.eye(nm.size)
    # Compute solution
    nm = ek @ nm + (ek-i) @ ki @ wr
    return na, nm
```

```python
def dis_fun(t, c):
    f = np.zeros_like(c)
    # Extract number densities
    n = c * const.UNA
    na, nm = n[0:1], n[1:]
    # Construct source terms
    wd = kd*nm*na - kr*na**3
    # Compose source terms
    f[0] = 2*np.sum(wd)
    f[1:] = - wd
    return f / const.UNA

def dis_int(t, na, nm):
    n0 = np.concatenate([na, nm])
    res = sp.integrate.solve_ivp(
        fun=dis_fun,
        t_span=[0,t],
        y0=n0/const.UNA,
        method='LSODA',
        atol=1e-9,
        rtol=1e-7
    )
    n = res.y[:,-1] * const.UNA
    return n[0:1], n[1:]
```

> Operator splitting

```python
def integrate(t, n0):
    y = [n0]
    na, nm = n0[0:1], n0[1:]
    for (i, ti) in enumerate(t[1:]):
        # Step 0: Define time step
        dt = ti - t[i]
        # Step 1: Integrate exc. for dt/2
        na, nm = exc_int(dt/2, na, nm)
        # na, nm = exc_dis_int(dt/2, na, nm)
        # Step 2: Integrate dis. for dt
        na, nm = dis_int(dt, na, nm)
        # Step 3: Integrate exc. for dt/2
        na, nm = exc_int(dt/2, na, nm)
        # na, nm = exc_dis_int(dt/2, na, nm)
        # Step 4: Store solution
        n = np.concatenate([na, nm])
        y.append(n)
    return np.vstack(y)
```
