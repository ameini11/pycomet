### Merge samples
Import libraries

```python
import os
import pandas as pd
```

Define inputs

```python
path_to_samples = "./../output/sampling/"
```

Merge samples

```python
samples = {}
for d in os.listdir(path_to_samples):
  d = f"{path_to_samples}/{d}/init_sols/"
  d = os.path.abspath(d)
  if os.path.isdir(d):
    for k in ("train", "valid", "test"):
      filename = f"{d}/{k}.csv"
      if os.path.exists(filename):
        df = pd.read_csv(filename)
        if (k not in samples):
          samples[k] = df
        else:
          samples[k] = pd.concat([samples[k], df], ignore_index=True, sort=False)
```

Save samples

```python
path_to_save = path_to_samples + "/samples/"
os.makedirs(path_to_save, exist_ok=True)
for (k, df) in samples.items():
  df.to_csv(path_to_save + f"/{k}.csv", float_format="%.8e", index=False)
```
