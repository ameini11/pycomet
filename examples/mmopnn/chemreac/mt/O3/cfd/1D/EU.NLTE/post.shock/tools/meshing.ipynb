### Meshing
Import libraries

```python
import os
import numpy as np
```

Define inputs

```python
log_scale = False
fac = 1.1
# With stretching factor
x_end = 1.
dx = 1.e-7
dx_min = x_end * 1.e-1
# With log-scale
nb_nodes = 500
limits = [dx, x_end]
```

Construct mesh

```python
if log_scale:
    l = np.log10(np.array(limits))
    x = np.logspace(l[0], l[1], num=nb_nodes-1, endpoint=True)
    x = np.concatenate([np.array([0.]), x]).reshape(-1,1)
else:
    x = [0., dx]
    while x[-1] < x_end:
        dx_old = x[-1] - x[-2]
        dx_new = dx_old * fac
        dx_new = min(dx_new, dx_min)
        x_new = min(x[-1] + dx_new, x_end)
        x.append(x_new)
    x = np.array(x)
```

Write mesh and boundary files

```python
nb_nodes = len(x)
path = os.path.abspath(os.getcwd() + f"/../meshes/n{nb_nodes}/")
os.makedirs(path, exist_ok=True)
```

```python
f = open(path + "/nodes.dat", "w")
f.write(f"{nb_nodes} 1 1\n")
np.savetxt(f, x, delimiter=' ', fmt='%.10E')
f.close()
```

```python
f = open(path + "/bound.dat", "w")
f.write("2\n")
f.write("INLET\n")
f.write(f"1 1 1\n")
f.write("OUTLET\n")
f.write(f"2 {nb_nodes} {nb_nodes}\n")
f.close()
```
