```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

import matplotlib as mpl
from matplotlib import pyplot as plt
COLORS = mpl.rcParams['axes.prop_cycle'].by_key()['color']*10
```

```python
def plot_qoi(
    x,
    y,
    var_names=None,
    ylabel=None,
    figname='',
    xlim=[1.e-6,1.]
):
    plt.figure()

    # Plotting
    plt.plot(x, y)

    # Set up axes
    plt.xlabel(r'$x$ [m]')
    plt.ylabel(ylabel)
    plt.xscale('log')
    plt.xlim(xlim)
    plt.ticklabel_format(axis='y', style='sci', scilimits=(0,0))

    # Set up legend
    if var_names is None:
        n_vars = 1
    else:
        var_names = tf.nest.flatten(var_names)
        n_vars = len(var_names)
    if var_names is not None:
        plt.legend(
            [plt.plot([], [], c=COLORS[i])[0] for i in range(n_vars)],
            var_names
        )

    # Save figure
    plt.savefig(figname)
    plt.close()
            
```

```python
cwd  = os.getcwd()
file = cwd + '/output_shock/shock.dat'
cols = ['x','X_O','X_O2','u','Th','Tv','rho','p','nd','H','Mf']
```

```python
path_to_figs = cwd + '/figs/'
os.makedirs(path_to_figs, exist_ok=True)
```

```python
df = pd.DataFrame(data=np.loadtxt(file), columns=cols)
```

```python
plot_qoi(
    df['x'],
    df['u'],
    ylabel=r'$u$ [m/s]',
    figname=path_to_figs + '/velocity.pdf'
)
plot_qoi(
    df['x'],
    df['p'],
    ylabel=r'$p$ [Pa]',
    figname=path_to_figs + '/pressure.pdf'
)
plot_qoi(
    df['x'],
    df[['Th','Tv']],
    ylabel=r'$T$ [K]',
    var_names=[r'$T_{h}$', r'$T_{v}$'],
    figname=path_to_figs + '/temperature.pdf'
)
plot_qoi(
    df['x'],
    df[['X_O2','X_O']],
    ylabel=r'$X$',
    var_names=[r'O$_2$', 'O'],
    figname=path_to_figs + '/molar_frac.pdf'
)
```
