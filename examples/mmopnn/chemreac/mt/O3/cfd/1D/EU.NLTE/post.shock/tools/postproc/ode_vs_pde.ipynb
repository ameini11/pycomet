```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
model = 'thermolib'
scheme = 'strang'
if scheme == 'implicit':
    file_name = 'flowfield_EU_NLTE.dat'
else:
    file_name = 'flowfield_EU_NLTE_100000_0.829749E-05s.dat'
xlim = [1.e-7,1.]
```

```python
cwd = os.getcwd()
data = {
    'ode': {
        'file': os.path.abspath(cwd + '/../../ode/output_shock/shock.dat'),
        'cols': ['x','X_O','X_O2','u','Th','Tv','rho','p','nd','H','Mf']
    },
    'pde': {
        'file': os.path.abspath(cwd + f'/../../pde/{model}/{scheme}/output/{file_name}'),
        'cols': ['x','X_O','X_O2','Th','Tv','rho','p','H','Mf','u']
    }
}
```

```python
dfs = {k: pd.DataFrame(data=np.loadtxt(v['file']), columns=v['cols']) for k, v in data.items()}
```

```python
for k in data.keys():
    plt.semilogx(dfs[k]['x'], dfs[k]['rho'])
plt.xlim(xlim)
```

```python
for k in data.keys():
    plt.semilogx(dfs[k]['x'], dfs[k]['u'])
plt.xlim(xlim)
```

```python
for k in data.keys():
    plt.semilogx(dfs[k]['x'], dfs[k]['X_O'])
    plt.semilogx(dfs[k]['x'], dfs[k]['X_O2'], '--')
plt.xlim(xlim)
```

```python
for k in data.keys():
    plt.semilogx(dfs[k]['x'], dfs[k]['Th'])
    plt.semilogx(dfs[k]['x'], dfs[k]['Tv'], '--')
plt.xlim(xlim)
```

```python
for k in data.keys():
    plt.semilogx(dfs[k]['x'], dfs[k]['Tv'])
plt.xlim(xlim)
```
