### Meshing
Import libraries

```python
import os
import numpy as np
```

Define inputs

```python
nb_nodes = 501
limits = [0., 0.01]
```

Construct mesh

```python
x = np.linspace(limits[0], limits[1], num=nb_nodes, endpoint=False)
x_far = np.linspace(limits[1], limits[1]*5, num=int(nb_nodes*0.4), endpoint=True)
x = np.concatenate([x,x_far])
```

Write mesh and boundary files

```python
nb_nodes = len(x)
path = os.path.abspath(os.getcwd() + f"/../meshes/n{nb_nodes}/")
os.makedirs(path, exist_ok=True)
```

```python
f = open(path + "/nodes.dat", "w")
f.write(f"{nb_nodes} 1 1\n")
np.savetxt(f, x, delimiter=' ', fmt='%.10E')
f.close()
```

```python
f = open(path + "/bound.dat", "w")
f.write("2\n")
f.write("INLET\n")
f.write(f"1 1 1\n")
f.write("OUTLET\n")
f.write(f"2 {nb_nodes} {nb_nodes}\n")
f.close()
```
