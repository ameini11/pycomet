Import libraries

```python
import os
import sys
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
sys.path.append(os.path.abspath('./../functions'))

%load_ext kernel_manager
```

Define inputs

```python
# Thermodynamics: mt - cg
qoi = {
    'names': ['Y_O', 'Y_O2', 'Th', 'Tv'],
    'labels': {
        'Y_O2': (r"$Y_{\text{O}_2}$",),
        'Y_O': (r"$Y_{\text{O}}$",),
        'Th': (r"$T$", "[K]"),
        'Tv': (r"$T_{\text{v}}$", "[K]")
    }
}
# Thermodynamics: mt - cg
species = ['O', 'O2']
nb_bins = 27
phys = ['rho','p','nd','H','Mf']
# Working case directory
path_to_case = "./../../"
```

```python
skip_flags = {
    'profiles': False
}
```

Read PLATO shock solution

```python
data = {}
for k in ('mt','cg'):
    if k == 'mt':
        s = [f'X_{s}' for s in species]
        t = ['Th', 'Tv']
        cols = ['x'] + [f'X_{s}' for s in species] + ['u','Th','Tv'] + phys
    else:
        s = [f'X_{species[0]}'] + [f'X_{species[1]}_{b+1}' for b in range(nb_bins)]
        t = ['Th']
    cols = tf.nest.flatten(['x', s, 'u', t, phys])
    d = np.loadtxt(path_to_case + f'/ode/{k}/output_shock/shock.dat')
    data[k] = pd.DataFrame(d, columns=cols)
```

```python
from matplotlib import pyplot as plt

plt.semilogy(data['mt']['X_O'][500:2500])
plt.semilogy(data['cg']['X_O'][500:2500])
plt.show()
```
