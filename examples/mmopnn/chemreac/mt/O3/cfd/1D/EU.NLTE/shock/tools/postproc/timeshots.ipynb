```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cwd = os.getcwd()
models = ['surrogate', 'thermolib']
scheme = 'strang'
nb_nodes = 701
prefix = {model: os.path.abspath(cwd + f'/../../pde/{model}/{scheme}/output/') + '/' for model in models}
iters = np.arange(0, 2001, 100)
files = {model: {str(i): prefix[model] + f'restart_EU_NLTE_{i}_0.dat' for i in iters} for model in models}
cols = ['rho_O','rho_O2','u','Th','Tv']
```

```python
dfs = {model: {k: pd.DataFrame(data=np.loadtxt(f), columns=cols) for k, f in files[model].items()} for model in models}
```

```python
path_to_mesh = os.path.abspath(os.getcwd() + f'/../../meshes/n{nb_nodes}/nodes.dat')
mesh = np.loadtxt(path_to_mesh, skiprows=1)
x = mesh[:-1] + 0.5*(mesh[1:] - mesh[:-1])
```

```python
for m in models:
    for k in files.keys():
        plt.plot(x, dfs[m]['2000']['Th'])
        # plt.plot(x, dfs[m][k]['Tv'], '--')
plt.xlim([0.003,0.008])
plt.ylabel('T [K]')
plt.xlabel('x [m]')
# plt.savefig(f'./{model}.temp.pdf')
```

```python
for m in models:
    for k in files.keys():
        plt.plot(x, dfs[m]['1000']['Tv'])
        # plt.plot(x, dfs[m][k]['Tv'], '--')
plt.xlim([0.003,0.008])
plt.ylabel('T [K]')
plt.xlabel('x [m]')
plt.savefig(f'./{model}.temp.pdf')
```

```python
for m in models:
    for k in files.keys():
        plt.plot(x, dfs[m]['2000']['Tv'])
        # plt.plot(x, dfs[m][k]['Tv'], '--')
plt.xlim([0.003,0.008])
plt.ylabel('T [K]')
plt.xlabel('x [m]')
plt.savefig(f'./{model}.temp.pdf')
```

```python
for m in models:
    for k in files.keys():
        plt.plot(x, dfs[m]['2000']['Th'])
        # plt.plot(x, dfs[m][k]['Tv'], '--')
plt.xlim([0.003,0.02])
plt.ylabel('T [K]')
plt.xlabel('x [m]')
plt.savefig(f'./{model}.temp.pdf')
```

```python
for k in files.keys():
    rho = np.sum(dfs[k][['rho_O','rho_O2']].to_numpy(), axis=1)
    plt.plot(x, dfs[k]['rho_O']/rho)
    plt.plot(x, dfs[k]['rho_O2']/rho, '--')
plt.xlim([0.003,0.02])
plt.ylabel('Y')
plt.xlabel('x [m]')
plt.savefig(f'./{model}.mass.pdf')
```
