```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cwd = os.getcwd()
model = 'thermolib'
scheme = 'strang'
nb_nodes = 701
prefix = os.path.abspath(cwd + f'/../../pde/{model}/{scheme}/output/') + '/'
iters = np.arange(500,5001,100)
files = {str(i): prefix + f'restart_EU_NLTE_{i}_0.dat' for i in iters}
cols = ['rho_O','rho_O2','u','Th','Tv']
```

```python
dfs = {k: pd.DataFrame(data=np.loadtxt(f), columns=cols) for k, f in files.items()}
```

```python
path_to_mesh = os.path.abspath(os.getcwd() + f'/../../meshes/n{nb_nodes}/nodes.dat')
mesh = np.loadtxt(path_to_mesh, skiprows=1)
x = mesh[:-1] + 0.5*(mesh[1:] - mesh[:-1])
```

```python
for k in files.keys():
    plt.plot(x, dfs[k]['u'])
plt.xlim([0.,0.05])
```

```python
for k in files.keys():
    rho = np.sum(dfs[k][['rho_O','rho_O2']].to_numpy(), axis=1)
    plt.plot(x, dfs[k]['rho_O']/rho)
    plt.plot(x, dfs[k]['rho_O2']/rho, '--')
plt.xlim([0.,0.05])
```

```python
for k in files.keys():
    plt.plot(x, dfs[k]['Th'])
    plt.plot(x, dfs[k]['Tv'], '--')
plt.xlim([0.,0.05])
```
