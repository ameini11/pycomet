```python
import numpy as np
import tensorflow as tf
import pandas as pd

from matplotlib import pyplot as plt
```

```python
path = '/home/zanardi/Workspace/ML/PyCOMET/chemreac/mt/O3/run.01.T/'
model_path = path + '/saving/mt_O3_mmopnn/SavedModel/'
data_path = path + '/data/test/case.010.csv'
```

```python
# Load Model
model = tf.keras.models.load_model(filepath=model_path)
```

```python
cols = {
    'x': [[0,1,2,3],[4]],
    'y': [[5,6],[7,8]]
}
```

```python
X = pd.read_csv(data_path).to_numpy(dtype=np.float32)
x = [np.take(X, i, axis=1) for i in cols['x']]
y = [np.take(X, i, axis=1) for i in cols['y']]
```

```python
y_pred = model(x)
```

```python
tf.nest.map_structure(np.shape, x)
```

```python
plt.semilogx(x[1], y[1])
plt.semilogx(x[1], y_pred[1], '--')
```
