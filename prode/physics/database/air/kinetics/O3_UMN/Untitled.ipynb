```python
import pandas as pd

grouping = 'RVE27'

arr = grouping + '/arrhenius_coeff.csv'
reac =  grouping + '/reactions.csv'

df_arr = pd.read_csv(arr)
df_reac = pd.read_csv(reac)
```

```python
df_arr = df_arr.rename({'Ea': 'Ta'}, axis='columns')

df_arr = df_arr.assign(Ta=df_reac['Ta'])
df_arr = df_arr.assign(A=df_reac['A'])
df_arr = df_arr.assign(n=df_reac['beta'])

df_arr.to_csv(grouping + '/arrhenius_coeff2.csv', index=False)
```
