```python
import numpy as np
import dill as pickle
import tensorflow as tf
import sys
sys.path.append("/home/zanardi/Codes/ML/PyCOMET/pycomet/")

from pycomet.networks.chemkin.plasma_cr.meno.crmodel import CRModel
```

```python
path_to_rates = "./rates2.p"
path_to_species = "/home/zanardi/Codes/ML/ROMAr/romar/examples/database/species/"
species = ("em", "Ar", "Arp")
```

```python
with open(path_to_rates, "rb") as file:
  rates = pickle.load(file)
```

```python
model = CRModel(
  species={k: path_to_species + f"/{k}.json" for k in species},
  processes=path_to_rates,
  use_corr=True,
  scale_corr=False,
  species_order=species
)
```

```python
Th = np.random.rand(100,1)*4e4 + 300
Te = np.random.rand(100,1)*2e4 + 300
f = np.random.rand(100,32,4)
```

```python
rates = model.compute_rates(Th, Te, f)
```
