```python
import tensorflow as tf
import numpy as np
import sys

sys.path.append("/home/zanardi/Codes/ML/PyCOMET/pycomet")
from pycomet.networks.meno.blocks import LinPROM
```

```python
prefix = "/home/zanardi/Workspace/AirDatabase/ThermoChemModels/StS/cr_argon/database/mixture/rom_data/"
linae = LinPROM(
  phi=prefix+"phi.txt",
  psi=prefix+"psi.txt",
  xref=prefix+"xref.txt",
  xscale=prefix+"xscale.txt",
  mask=prefix+"mask.txt",
  z_dim=9
)
linae.build()
```

```python
n = 34
x = tf.random.uniform((n,), dtype="float64")
z = tf.linalg.matvec(tf.transpose(linae.encoder), x)
x[0]-z[-1]
```

```python
# Ions indices
iions = [32,33]
# Mask vector
imask = np.zeros(34)
imask[iions] = 1.0
imask = tf.constant(imask, dtype="float64")
# Decoder and refernce values for ions
idec = tf.linalg.matvec(linae.decoder, imask)
print(idec.shape)
iref1 = tf.tensordot(imask, linae.xref, axes=1)
iref2 = tf.math.reduce_sum(imask*linae.xref)
print(iref1-iref2)
```
