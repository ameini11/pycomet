```python
import tensorflow as tf
import numpy as np
```

```python
n = 20

a = tf.random.uniform((10, n, n))
print('a', a.shape)
d = tf.random.uniform((n, 7))
print('d', d.shape)
e = tf.random.uniform((6, n))
print('e', e.shape)

ah = tf.linalg.matmul(a, d)
print('ah', ah.shape)
ah = tf.linalg.matmul(e, ah)
print('ah', ah.shape)
```
