```python
import numpy as np
import tensorflow as tf
import pandas as pd
```

```python
path = '/home/zanardi/Codes/ML/PyCOMET/pycomet/examples/pinn/forward/msd/'
model_path = path + '/saving/MassSpringDamper_FFNN/model.tflite'
data_path = path + '/data/test/case_001.csv'
```

```python
# Load the TFLite model and allocate tensors.
interpreter = tf.lite.Interpreter(model_path=path)
interpreter.allocate_tensors()

# Get input and output tensors.
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()
```

```python
X = pd.read_csv(data_path).to_numpy(dtype=np.float32)
x = [np.take(X, [0,1], axis=1), np.take(X, [2], axis=1)]
y = np.take(X, [3,4], axis=1)
```

```python
for i, inp in enumerate(input_details):
    interpreter.resize_tensor_input(inp['index'], x[i].shape, strict=True)
    interpreter.allocate_tensors()
    interpreter.set_tensor(inp['index'], x[i])
interpreter.invoke()
y_pred = interpreter.get_tensor(output_details[0]['index'])
```

```python
from matplotlib import pyplot as plt
plt.plot(x[1], y)
plt.plot(x[1], y_pred, '^-')
```
