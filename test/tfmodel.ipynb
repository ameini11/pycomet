```python
import numpy as np
import tensorflow as tf
import pandas as pd

from matplotlib import pyplot as plt
```

```python
path = '/home/zanardi/Codes/ML/PyCOMET/pycomet/examples/pinn/forward/msd/'
model_path = path + '/saving/MassSpringDamper_FFNN/SavedModel'
data_path = path + '/data/test/case_0.csv'
```

```python
# Load Model
model = tf.keras.models.load_model(filepath=model_path)
```

```python
X = pd.read_csv(data_path).to_numpy(dtype=np.float32)
x = np.take(X, [0], axis=1)
y = np.take(X, [1,2], axis=1)
```

```python
y_pred = model(x)
```

```python
plt.plot(x, y)
plt.plot(x, y_pred, '^-')
```
