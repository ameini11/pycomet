```python
import numpy as np
import tensorflow as tf
import pandas as pd

from matplotlib import pyplot as plt
```

```python
path = '/home/zanardi/Codes/ML/PyCOMET/pycomet/examples/opnn/msd/'
model_path = path + '/saving/msd_flexopnn/SavedModel'
data_path = path + '/data/test/case_001.csv'
```

```python
# Load Model
model = tf.keras.models.load_model(filepath=model_path)
```

```python
columns = {
    'x': [[0,1],[2]],
    'y': [3,4]
}
```

```python
X = pd.read_csv(data_path).to_numpy(dtype=np.float32)
x = [np.take(X, i, axis=1) for i in columns['x']]
y = np.take(X, columns['y'], axis=1)
```

```python
y_pred = model(x)
```

```python
plt.plot(x[1], y)
plt.plot(x[1], y_pred, '^-')
```
